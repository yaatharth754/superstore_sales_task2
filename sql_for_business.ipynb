# What are the Top 5 Products by Revenue?

```python
query = """
SELECT "Product Name", SUM(Sales) AS Total_Revenue
FROM sales
GROUP BY "Product Name"
ORDER BY Total_Revenue DESC
LIMIT 5;
"""

pd.read_sql(query, conn)
```

# Monthly Sales Trend

```python
query2 = """
SELECT strftime('%Y', "Order Date") AS Order_Year,
       strftime('%m', "Order Date") AS Order_Month,
       SUM(Sales) AS Monthly_Sales
FROM sales
GROUP BY Order_Year, Order_Month
ORDER BY Order_Year, Order_Month;
"""

pd.read_sql(query2, conn)
```

# Revenue by Region

```python
query3 = """
SELECT Region, SUM(Sales) AS Total_Sales
FROM sales
GROUP BY Region
ORDER BY Total_Sales DESC;
"""

pd.read_sql(query3, conn)
```

# Revenue by Customer Segment

```python
query4 = """
SELECT Segment, SUM(Sales) AS Segment_Revenue
FROM sales
GROUP BY Segment
ORDER BY Segment_Revenue DESC;
"""

pd.read_sql(query4, conn)
```

# Average Shipping Time by Ship Mode

```python
query5 = """
SELECT "Ship Mode", AVG("Shipping Days") AS Avg_Shipping_Time
FROM sales
GROUP BY "Ship Mode"
ORDER BY Avg_Shipping_Time DESC;
"""

pd.read_sql(query5, conn)
```

# Top 10 Customers by Revenue

```python
query6 = """
SELECT "Customer Name", SUM(Sales) AS Customer_Revenue
FROM sales
GROUP BY "Customer Name"
ORDER BY Customer_Revenue DESC
LIMIT 10;
"""

pd.read_sql(query6, conn)
```
